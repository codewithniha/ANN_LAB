# Task 3: Addition & Subtraction

```python
marks_term1 = np.array([[90,90,90],
                       [98,98,98],
                       [100,100,100]])
marks_term2 = np.array([[95,95,95],
                       [98,98,98],
                       [95,95,95]])
combined_marks = marks_term1 + marks_term2
improved_marks = marks_term2 - marks_term1

improved_marks
```

```python
total_improvement = improved_marks.sum(axis=1)
print(total_improvement)
```

```python
print(f"student who improved highest is student at position: {np.argmax(total_improvement) + 1}")
```

# Task 4: Dot Product

```python
prices = np.array([50,50,100,200])
items = np.array([1,5,2,2])

bill = np.dot(prices,items)
bill
```

```python
prices = np.array([50,50,100,200])
qnty = np.array([[3,4,6,0],[9,6,9,10]])
bill = np.dot(prices,qnty)
```

conditions for dot product: 

   - for one dimensional vector: both arrays must have the same number of elements.

   - for two dimensional vector: The columns of the first array must match the rows of the second array.      ( rows, columns ) dot ( rows, columns )

# Task 9: Transpose

```python
employees_performance = np.array([[10,100,20,40,50],[40,50,60,70,80],[90,50,40,50,60]])
employees_performance
```

```python
employees_performance.shape
```

```python
employees_performance_single_trnspose = np.transpose(employees_performance)
employees_performance_single_trnspose.shape
```

```python
employees_performance_double_trnspose = np.transpose(employees_performance_single_trnspose)
employees_performance_double_trnspose.shape
```

```python
if(employees_performance.shape == employees_performance_double_trnspose.shape):
        print("transposing the matrix twice gives back the original matrix verified")
else:
        print("transposing the matrix twice doesn't gives back the original matrix verified")
```

```python
matrix_1 = np.random.rand(4,2)
matrix_2 = np.random.rand(3,2)
```
